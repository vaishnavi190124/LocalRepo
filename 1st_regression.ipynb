```python
import numpy as np
#data
X=np.array([1,2,3,4,5],dtype=float)
Y=np.array([2,4,5,4,5],dtype=float)
```

```python
#initial values
b0=0
b1=0
learning_rate=0.01
epochs=1000
n=len(X)
```

```python
epochs=1000
#gradient descent
for i in range(epochs):
  #predictions
  Y_pred=b0+b1*X
  #calculate gradients
  db0 = (-2/n) * np.sum(Y - Y_pred)
  db1 = (-2/n) * np.sum(X * (Y - Y_pred))
  # update parameters
  b0 = b0 - learning_rate * db0
  b1 = b1 - learning_rate * db1
```

```python
print("b0=",b0)
print("b1=",b1)
print("Regression Equation:")
print("Y=",b0,"+",b1,"X")
```
